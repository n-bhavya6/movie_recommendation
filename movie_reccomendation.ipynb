```python
import pandas as pd
```

```python
import io

csv_data = """movie_id,title,genre,rating
1,Inception,Sci-Fi,8.8
2,Titanic,Romance,7.8
3,Avengers Endgame,Action,8.4
4,The Dark Knight,Action,9.0
5,Interstellar,Sci-Fi,8.6"""

df = pd.read_csv(io.StringIO(csv_data))
print(df)
```

```python
import io
import pandas as pd

user_ratings_csv = """user_id,movie_id,rating
1,1,5
1,2,3
2,1,4
2,3,5
3,2,2
3,4,5"""

user_ratings_df = pd.read_csv(io.StringIO(user_ratings_csv))
print(user_ratings_df)
```

```python
merged_df = pd.merge(df, user_ratings_df, on='movie_id')

average_ratings_per_genre = merged_df.groupby('genre')['rating_y'].mean().reset_index()

print("Average rating per genre:")
print(average_ratings_per_genre)
```

```python
average_rating_per_movie = merged_df.groupby('title')['rating_y'].mean().reset_index()
top_rated_movie = average_rating_per_movie.sort_values(by='rating_y', ascending=False).iloc[0]

print("Top-rated movie based on user ratings:")
print(top_rated_movie)
```

```python
import pandas as pd

# movies = pd.read_csv('movies.csv') # This line caused the error
# ratings = pd.read_csv('ratings.csv') # This line would also cause an error

# Use the already loaded DataFrames 'df' and 'user_ratings_df'
# assuming df contains movies data and user_ratings_df contains ratings data.
# If the user intends to rename them, they can assign them.
movies_df = df # Renaming for clarity
ratings_df = user_ratings_df # Renaming for clarity

print("Movies Data:")
print(movies_df)

print("\nRatings Data:")
print(ratings_df)
```

```python
user_movie_matrix = merged_df.pivot_table(index='user_id', columns='title', values='rating_y')
print(user_movie_matrix)
```

```python
movie_correlation_matrix = user_movie_matrix.corr()
print("Movie Correlation Matrix:")
print(movie_correlation_matrix)
```
